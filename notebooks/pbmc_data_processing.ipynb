```python
cite_seq = ad.read_h5ad("data/pbmc_citeseq/cite_seq.h5ad")
rna = ad.read_h5ad("data/pbmc_citeseq/rna_final.h5ad")

valid_cells = rna.obs['celltype.l1'] != 'other'
rna = rna[valid_cells].copy()
cite_seq = cite_seq[valid_cells].copy()

rna.var['mt'] = rna.var_names.str.startswith('MT-')
rna = rna[:, ~rna.var['mt']].copy()

sc.pp.normalize_total(rna, target_sum=1e4)
sc.pp.log1p(rna)
sc.pp.highly_variable_genes(rna, n_top_genes=2000, subset=True)

# 3. Preprocess Protein (Muon CLR)
mu.prot.pp.clr(cite_seq)

# 4. Save
rna.write("data/pbmc_citeseq/rna_final_processed.h5ad")
cite_seq.write("data/pbmc_citeseq/cite_seq_processed.h5ad")
```
